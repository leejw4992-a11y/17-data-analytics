```python
import pandas as pd
```

```python
dict_data = {
    "c1" : ["a","a","b","a","b"],
    "c2" : [1,1,1,2,2],
    "c3" : [1,1,2,2,2]
}

df= pd.DataFrame(dict_data)
df
```

```python
df_dup_first = df.duplicated()
df_dup_first
```

```python
df_dup_last = df.duplicated(keep="last")
df_dup_last
```

```python
df_dup_false = df.duplicated(keep=False)
df_dup_false
```

```python
df
```

```python
col_dup = df["c2"].duplicated()
col_dup
```
