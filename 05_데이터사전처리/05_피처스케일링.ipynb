```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./data/auto-mpg.csv",header=None)

df.columns = [
    "mpg",'cylinders','displacement',"horsepower","weight","acceleration","model year","origin","name"
]

df["horsepower"] = df["horsepower"].replace("?", np.nan)
df = df.dropna(subset=["horsepower"], axis=0)
df["horsepower"] = df["horsepower"].astype("float")

df["horsepower"].describe()
```

```python
# min-max scailing

df["horsepower_minmax"] = (df["horsepower"]-df["horsepower"].min()) / df["horsepower"].max() - df["horsepower"].min()
```

```python
df.head()
```

```python
df["horsepower_minmax"].describe()
```

```python
#사이킷런에서 min-max 스케일링

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df["hs_minmax2"] = scaler.fit_transform(df[["horsepower"]])
df["hs_minmax2"].head()
```

```python
df["hs_minmax2"].describe()
```

```python
#standard Scaling
## 평균은 0, 표준편차는 1

df["horsepower_standard"] = (df["horsepower"] - df["horsepower"].mean()) / df["horsepower"].std()
print(df["horsepower_standard"].head())
print(df["horsepower_standard"].info())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df["hp_standard2"] = scaler.fit_transform(df[["horsepower"]])
df["hp_standard2"].head()
```
