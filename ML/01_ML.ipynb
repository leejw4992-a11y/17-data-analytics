```python
import sklearn # 사이킷런 전체 라이브러리 불러오기
# from sklearn import set_config
# set_config(display="text")


from sklearn.datasets import load_iris # 사이킷 런 안에 데이타셋에서 아이리스 데이터 불러오기
from sklearn.tree import DecisionTreeClassifier # 사이킷런 안에 트리 안에서 결정트리모델 불러오기
from sklearn.model_selection import train_test_split # 사이킷런 안에 모델 셀렉션에서 데이터를 학습용/테스트용 나누는 도구 불러오기
```

```python
iris = load_iris() # 불러온 아이리스데이터를 아이리스 변수에 할당하기
import pandas as pd # 판다스 pd로 가져다 쓰기

# Iris 데이터 셋에서 피쳐(X 값)
iris_data = iris.data # 아이리스 데이터에서 X 특징

# Iris 데이터 셋에서 Y값
iris_label = iris.target # 아이리스 데이터에서 Y 정답 

# print(iris_label) #꽃 품종 을 0,1,2로 나타냄
# print(iris.target_names) #꽃 종류의 이름 3종


iris_df = pd.DataFrame(data = iris_data, columns=iris.feature_names)  # X 특징 데이터를 표로 만들고 열은 꽃 특징 
iris_df["label"] = iris.target # 아이리스 Y정답을 label열로 추가해서 할당하기
iris_df.head() # 상위 5개 찍어보기
```

```python
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label,test_size= 0.2, random_state = 11)
# 아이리스 데이터를 두덩어리로 나누기 train_test_split을 써서 test_size는 20%, 학습용은 80%의미 / 섞는 기준 11개
# X_train/y_train (학습용 특징/정답)
# X_test/y_test (테스트용 특징/정답)

# 나누는 이유 
# 학습시킨 데이터로 테스트를 보면 안됨  -> 다른말로 처음보는 데이터도 잘 맞히나 확인하는것임
```

```python
dt_clf = DecisionTreeClassifier(random_state=11) 
# 결정트리모델을 만들어 df_clf에 담고 random_state=11로 트리가 매번 같은 결과를 내게 고정함.
dt_clf.fit(X_train, y_train);
# 학습용 특징과 정답(X_train, y_train)을 가지고 학습 시킴. -> 이렇게 되면 트리가 어떤 꽃 크기면 어떤 품종인지에 대한 규칙을 배움.
```

```python
pred = dt_clf.predict(X_test)
# print(pred)
# X_train과 y_train을 가지고 학습시킨 dt_clf변수에다가
# X_test를가지고 predict를 사용하여 예측값을 계산시켜 pred라는 변수에 담음.
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, pred))
```
