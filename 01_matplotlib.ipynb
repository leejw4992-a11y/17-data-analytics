```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd. read_excel("./data/시도별_전출입_인구수.xlsx")

df.head()
df.info()
```

```python
df.head(10)
```

```python
#결측값(누락값)을 앞데이터로 채움.

df = df.ffill()
df.head(10)
```

```python
df.head(1)
```

```python
mask = (df["전출지별"] == "서울특별시") & (df["전입지별"] != "서울특별시")
df_seoul = df[mask]
df_seoul = df_seoul.drop(["전출지별"], axis=1)
df_seoul = df_seoul.rename({"전입지별": "전입지"},axis = 1)
df_seoul = df_seoul.set_index("전입지")
print(df_seoul)
```

```python
sr_one = df_seoul.loc["경기도"]
sr_one.head()
```

```python
plt.plot(sr_one.index, sr_one.values)
```

```python
plt.plot(sr_one)
plt.show()
```

```python
# 차트 제목, 축 이름 추가 , 선 스타일 지정

plt.plot(sr_one.index, sr_one.values, linestyle="--")
plt.title("서울 -> 경기 인구 이동")

plt.xlabel("기간")
plt.ylabel("이동 인구수")

plt.show()
```

```python
# 한글 폰트 오류 해결

from matplotlib import font_manager, rc
font_path = "./data/malgun.ttf"
font_name = font_manager.FontProperties(fname=font_path).get_name()
rc('font',family=font_name)
```

```python
plt.plot(sr_one.index, sr_one.values, linestyle="--")
plt.title("서울 -> 경기 인구 이동")

plt.xlabel("기간")
plt.ylabel("이동 인구수")

plt.show()
```

```python
plt.style.use("ggplot")

plt.figure(figsize=(14,5))

plt.xticks(size=10, rotation='vertical')
plt.plot (sr_one.index, sr_one.values,
    marker = "O",
    markerfacecolor='red',
    markeredgecolor='blue',
    markeredgewidth='2',
    markersize=10
)

plt.title("서울 -> 경기 인구 이동")

plt.xlabel("기간",size=20)
plt.ylabel("이동 인구수",size=20)
plt.legend(labels=['서울->경기'],loc='best',fontsize=15)

plt.show
```

```python
import pandas as pd
import matplotlib.pyplot as plt


samsung_revenue = pd.read_csv("./data/삼성전자_분기별_매출액.csv")
samsung_revenue = samsung_revenue.sort_values("quarter")
samsung_revenue
```

```python
fig = plt.figure()
print(fig)
```

```python
fig, axe = plt.subplots()
fig
axe
```

```python
fig,axe = plt.subplots(figsize=(6,3))
axe.plot(samsung_revenue["quarter"],samsung_revenue['value'])
plt.show()
```

```python
fig,axes = plt.subplots(1,2)
print(fig)
print(axes)
```

```python
fig,axes = plt.subplots(1,2,figsize=(12,2))
axes[0].plot(samsung_revenue["quarter"],samsung_revenue['value'])
samsung_revenue['value'].plot(ax=axes[1])
plt.show()
```
