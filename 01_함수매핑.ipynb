```python
# 개별 원소의 함수 매핑

# Series객체.apply(매핑함수)

import seaborn as sns
import pandas as pd
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()

df = titanic.loc[:, ["age", "fare"]]
df.head(3)
```

```python
def add_10(n):
    return n+10
def add_two_obj(a,b):
    return a+b

print(add_10(10))
print(add_two_obj(10,10))
```

```python
sr3 = df["age"].apply(add_two_obj, b=10)
sr3.head()
```

```python
sr4 = df["age"].apply(lambda a,b : a + b, b=10)
sr4.head()
```

```python
# 데이터프레임객체.apply(매핑함수, axis=0 or 1)
df.head()
```

```python
def calculate_stats(col):
    max_val = col.max()
    min_val = col.min()
    mean_val = col.mean()
    median_val = col.median()

    df = pd.Series([
        max_val,
        min_val,
        mean_val,
        median_val
    ],index=["max","min","mean","median"])

    return df

result_df = df.apply(calculate_stats, axis = 0)
result_df.head()
```

```python
result_sr = df.apply(lambda x : x.max() - x.min(), axis = 0) #axis 는 세로형태 열로 취급한다는 의미 / 여러가지데이터를 하나의 수로 표현하는 집계함수
result_sr
```

```python
df.head(2)
```

```python
# 데이터프레임객체.apply(매핑함수, axis=1) ->행이 매핑함수의 인수로 들어간다.

def calculate_diff_avg(row):
    diff = row.max() - row.min()
    avg = row.mean()
    return pd.Series([diff, avg], index = ["차이","평균"])

result_df2 = df.apply(calculate_diff_avg, axis = 1)
result_df2
```

```python
result_df3 = df.apply(
    lambda row, multiplier: pd.Series(
        [(row.max() - row.min()) * multiplier, row.mean()],
        index=["차이", "평균"]
    ), multiplier=2, axis=1)


result_df3
```
