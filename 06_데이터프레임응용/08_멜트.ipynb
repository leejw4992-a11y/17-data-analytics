```python
# DataFrame객체.melt(id_vars=[],value_vars="")

import pandas as pd
```

```python
df = pd.DataFrame({
    "name":["John","Mary"],
    "city":["New York","Los Angeles"],
    "Age" : [28,32],
    "Salary": [50000,60000]
})
df
```

```python
df_melted = df.melt(
    id_vars = ["name","city"],
    var_name = "Attribue",
    value_name = "value"
)

df_melted
```

```python
df_wide = pd.DataFrame({
    "company": ["Google", "Yahoo"],
    "Income_2019": [100000, 50000],
    "Income_2020": [120000, 55000],
    "Expense_2019" : [50000, 30000],
    "Expense_2020" : [60000, 35000],
})
df_wide
```

```python
df_long = pd.wide_to_long(
    df_wide,
    stubnames=["Income","Expense"],
    i="Company",
    j="Year",
    sep="_",
    suffix="\d+"
)

df_long = df_long.reset_index()
df_long
```
