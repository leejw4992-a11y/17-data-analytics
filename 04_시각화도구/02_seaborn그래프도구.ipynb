```python
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.info()
```

```python
import matplotlib.pyplot as plt

sns.set_style("darkgrid")

fig, axes = plt.subplots(1,2,figsize=(15,5))

sns.regplot(
    x="age",
    y="fare",
    data=titanic,
    ax=axes[0]
)
sns.regplot(
    x="age",
    y="fare",
    data=titanic,
    ax=axes[1],
    fit_reg=False
)

plt.show()
```

```python
# 히스토그램 / 커널 밀도그래프

fig,axes = plt.subplot_mosaic([
    ["top_left","top_right"],
    ["middle_left","middle_right"],
    ["bottom","bottom"]
])

sns.histplot(
    x = "age",
    data = titanic,
    ax = axes["top_left"]
)
#2 hue 옵션 추가
sns.histplot(
    x = "age",
    hue = "survived",
    data = titanic,
    ax=axes['top_right']
)

# dodge 적용
sns.histplot(
    x="age",
    hue="survived",
    data = titanic,
    ax=axes['middle_left'],
    multiple= "dodge"
)

sns.histplot(
    x="age",
    hue="survived",
    data = titanic,
    ax=axes["middle_right"],
    multiple="stack"

)

sns.histplot(
    x="age",
    hue="survived",
    data=titanic,
    ax=axes['bottom'],
    multiple="fill",
    bins=10
)

fig.suptitle("Titanic=Age Distribution")

axes["top_left"].set_title("HIistogram")
axes["top_right"].set_title("HIistogram (hue)")
axes["middle_left"].set_title("HIistogram (multiple-dodge)")
axes["middle_right"].set_title("HIistogram (multiple-stack)")
axes["bottom"].set_title("HIistogram (multiple-fill)")
plt.show()
```

```python
fig,axes = plt.subplot_mosaic([
    ["top_left","top_center","right"],
    ["bottom_left","bottom_center","right"],
], figsize=(15,6),constrained_layout=True)

sns.kdeplot(x="age",data=titanic,ax=axes["top_left"])

sns.kdeplot(x="age",hue="survived",data=titanic,ax=axes["bottom_left"])

sns.kdeplot(x="age",hue="survived",data = titanic, fill=True, ax=axes["top_center"])

sns.kdeplot(x="age",hue="survived",data = titanic, multiple="stack", ax=axes["bottom_center"])

sns.kdeplot(x="age", hue="survived", multiple="fill", bw_adjust=1.0, data = titanic, ax=axes["right"])


fig.suptitle("Titanic = Age Distribution")

axes["top_left"].set_title("KDE")
axes["bottom_left"].set_title("KDE (hue)")
axes["top_center"].set_title("KDE (fill=True)")
axes["bottom_center"].set_title("KDE (multiple-stack)")
axes["right"].set_title("KDE (multiple - fill)")
```

```python
# 히트맵

sns.set_style("darkgrid")

table = titanic.pivot_table(
    index=["sex"],
    columns=["class"],
    aggfunc="size",
    observed=False
)

table
```

```python
sns.heatmap(
    table,
    annot=True, fmt="d",
    linewidth=5,
    cbar=True
    
)

plt.show()
```

```python
sns.set_style("whitegrid")
fig, axes = plt.subplots(1,2,figsize=(15,5))

sns.stripplot(
    x="class",
    y="age",
    data=titanic,
    ax=axes[0]
)
sns.stripplot(
    x="class",
    y="age",
    data=titanic,
    ax=axes[1],
    hue="class",
    size=4
)
```

```python
# 막대그래프
sns.set_style("ticks")

fig,axes = plt.subplots(2,2,figsize=(10,6), constrained_layout=True)

sns.barplot(
    x="sex",
    y="survived",
    data=titanic,
    ax = axes [0,0]
)
sns.barplot(
    x="sex",
    y="survived",
    data=titanic,
    ax = axes [0,1],
    errorbar=None
)

plt.show()
```

```python
#박스 플롯/바이올린 그래프

fig,axes = plt.subplots(2,2,figsize=(15,10))

sns.boxplot(
    x="alive",
    y="age",
    data=titanic,
    ax=axes[0,0]
)
sns.boxplot(
    x="alive",
    y="age",
    hue='sex',
    data=titanic,
    ax=axes[0,1]
)

sns.violinplot(
    x="alive",
    y="age",
    data=titanic,
    ax=axes[1,0]
)
sns.violinplot(
    x="alive",
    y="age",
    data=titanic,
    hue="sex",
    ax=axes[1,1]
)
```
