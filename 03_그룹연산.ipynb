```python
#그룹 객체 만들기

# DataFrame객체.groupby(기준이되는열)

import pandas as pd
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age","sex","class","fare","survived"]]
df.head()
```

```python
df["class"].value_counts()
df["class"].nunique()
df["class"].unique()
```

```python
grouped = df.groupby("class", observed=True)
grouped
```

```python
for key , group in grouped:
    print("* key * : ", key)
    print("* number * : ", len(group))
    print(group.head())
```

```python
average = grouped.mean(numeric_only=True) #숫자로 이루어진 데이터만 보여줘라 라는 말
average
```

```python
# 개별그룹 선택하기

group3 = grouped.get_group("Third")
group3
```

```python
group1 = grouped.get_group("First")
group1
```

```python
# 그룹 연산 메서드(적용-결합단계)

df.head()
```

```python
grouped = df.groupby(["class"], observed=True)
std_all = grouped.std(numeric_only=True)
std_all
```

```python
# aggregate() 매서드 데이터 집계: group객체.aggregate(매핑 함수)
# agg() 매서드 데이터 집계: group객체.agg(매핑 함수)

agg_mean = grouped.aggregate("mean",numeric_only=True)
agg_mean
```

```python
grouped.agg("mean",numeric_only=True)
```

```python
# group객체.agg([함수1, 함수2])
# group객체.agg({"열1": 함수1, "열2":함수2})

agg_all = grouped.agg(["min","max"])
agg_all
```

```python
agg_sep = grouped.agg({
    "fare":["min","max"],
    "age" : "mean"
})

agg_sep
```

```python
def min_max(x):
    return x.max() - x.min()


agg_minmax = grouped[["age","fare"]].agg(min_max)
agg_minmax
```

```python
#변환 transformation

df.head()

```

```python
grouped = df.groupby(["class"],observed=True)
grouped
```

```python
grouped["fare"].cumsum()
```

```python
df["fare_cumsum"] = grouped["fare"].cumsum() # 갚은 그룹끼리 누적값 계산하는것(cumsum).
df.head(10)
```

```python
# group객체/transform(매핑함수)

grouped[["fare"]].transform("cumsum")
```

```python
grouped[["age","survived"]].transform("mean") #age와 survived에 대한 평균값 
```

```python
def z_score(x):
    return (x - x.mean()) / x.std()

age_zscore = grouped["age"].transform(z_score)
age_zscore
```

```python
age_zscore2 = grouped["age"].transform(lambda x : (x - x.mean()) / x.std())
age_zscore2
```
