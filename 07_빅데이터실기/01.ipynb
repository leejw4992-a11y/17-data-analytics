````python
import pandas as pd
import numpy as np

# 데이터 로드
data_url = "https://raw.githubusercontent.com/YoungjinBD/dataset/main/airquality.csv"
airquality = pd.read_csv(data_url)
airquality.head()
````

## 문제 ① 결측치 처리 및 평균 비교

`airquality` 데이터에 대해서 **결측치가 가장 많은 변수**를 찾아서 해당 결측치를 **0으로 대치**하고,  
**결측치를 제외한 평균**과 **0으로 대치한 후의 평균**과의 차이를 구하시오.

> 출력: 두 평균의 차이 (절대값 여부는 풀이 방식에 따름)

````python
airquality.info()

airquality.isnull()
missing = airquality.isnull().sum().idxmax()
print(missing)

mean_before = airquality[missing].mean()
print(mean_before)

mean_after = airquality[missing].fillna(0).mean()
print(mean_after)

result = mean_before - mean_after
print(result)
result = round(result,3)
print(result)
````

## 문제 ② Min-Max / Z 정규화 평균 비교

`Wind` 변수에 대해서 **Min-Max 정규화**를 수행한 후 평균값과  
**Z 정규화(표준화)**를 수행한 후 평균값의 차이를 구하시오.

- Min-Max 정규화: $(x - min) / (max - min)$
- Z 정규화: $(x - mean) / std$

> 출력: 두 평균의 차이

````python
airquality 를 df로 바꿀것.

airquality["Wind"].head()

wind_min = airquality["Wind"].min()
wind_max = airquality["Wind"].max()
wind_mean = airquality["Wind"].mean()
wind_std = airquality["Wind"].std()

print(wind_min)
print(wind_max)
print(wind_mean)
print(wind_std)

airquality["minmax"] = (airquality["Wind"] - wind_min) / (wind_max - wind_min)
airquality.head()
airquality["zscore"] = (airquality["Wind"] - wind_mean) / wind_std
airquality.head()

result2 = airquality["minmax"].mean() - airquality["zscore"].mean()
result2 = round(result2,2)
result2
````
