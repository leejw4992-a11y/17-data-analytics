## 문제 ① Attrition 수치화 및 범주별 건수

해당 데이터에서 `Attrition`은 종속변수이다.  
Attrition의 값을 **수치형**으로 변환해 새로운 컬럼으로 추가하고,  
범주별 **레코드 수**를 구하시오.

- `Yes` → `1` (퇴사)
- `No` → `0` (퇴사하지 않음)

> 출력: Attrition(0/1) 범주별 건수

````python


# df.info()
df["Attrition"].head()

#1-1
df["a_le"] = df["Attrition"].map({"Yes":1, "No":0})
df["a_le"]

#1-2
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["a_le2"] = le.fit_transform(df["Attrition"])
df["a_le2"].head()
df["a_le2"].value_counts()





````
