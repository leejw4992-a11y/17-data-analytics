## 문제 ① 구급대 평균 이동시간

주어진 데이터를 이용하여 **이동시간**(출동시간과 도착시간의 차이)이  
평균적으로 **가장 오래 걸린 구급대**의 이동시간을 구하시오.

- 분 단위로 환산
- 반올림하여 **정수**로 출력

**데이터**: `data_6_1_1.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_1.csv
```

> 출력: 평균 이동시간(분, 정수)

````python


df = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_1.csv")
df.head()

# df.info()

df["출동시간"] = pd.to_datetime(df["출동시간"])
df["도착시간"] = pd.to_datetime(df["도착시간"])

df["이동시간"] = ((df["도착시간"] - df["출동시간"]).dt.total_seconds()/60).astype(int)
df["이동시간"]


mean_by = df.groupby("구급대")["이동시간"].mean()

print(mean_by.idxmax(), round(mean_by.max()))
````

## 문제 ② 교수 1인당 학생 수가 가장 많은 대학

주어진 데이터를 이용하여 **교수 1인당 맡고 있는 학생 수**가 가장 많은 대학교를 찾고,  
그 대학교의 **전체 교수 수**를 구하시오. (**정수**로 출력)

**데이터**: `data_6_1_2.csv`
```
https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_2.csv
```

> 출력: 해당 대학교의 전체 교수 수(정수)

````python

df2 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_2.csv")
df2.head()
````

````python


df2 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_2.csv")
df2.head()

학과 = df2.loc[:,'경영학과' : '영문학과'].columns
df2["학생수"] = df2[학과].sum(axis=1)
df2["학생수"]

df2["비율"] = df2["학생수"]/df2["교수"]
# df2["비율"]   

idx = df2["비율"].idxmax()
print(df2.loc[idx,"대학교명"], int(df2.loc[idx, "교수"]))
````

````python


df3 = pd.read_csv("https://raw.githubusercontent.com/YoungjinBD/dataset/main/data_6_1_3.csv")
df3.head()

# df3.info()
범죄 = df3.loc[:, '폭력범죄':'정치범죄'].columns
df3["범죄건수"] = df3[범죄].sum(axis=1)
df3["범죄건수"]

# df3["연도"] = df3["날짜"].str[0:4]
# df3.head()

df3["연도"] = pd.to_datetime(df3["날짜"], format="%Y년 %m월").dt.year
df3.head()

grouped = df3.groupby("연도")
grouped["범죄건수"].agg("mean").sort_values(ascending=False)
````
